```python
canvas = np.zeros((600, 800, 3), dtype=np.uint8)

h, w = img_flip.shape[:2]

x = (800 - w) // 2
y = (600 - h) // 2

if x >= 0 and y >= 0:
    canvas[y:y+h, x:x+w] = img_flip
else:
    resized = cv.resize(img_flip, (min(w, 800), min(h, 600)))
    rh, rw = resized.shape[:2]
    x = (800 - rw) // 2
    y = (600 - rh) // 2
    canvas[y:y+rh, x:x+rw] = resized

plt.figure(figsize=(10, 7))
plt.imshow(canvas)
plt.show()
```

```python
path = '/content/drive/MyDrive/PCVK/Gambar/KTP.jpg'

img_analisa = cv.imread(path)

gray = cv.cvtColor(img_analisa, cv.COLOR_BGR2GRAY)

rata_awal = np.mean(gray)

tinggi, lebar = gray.shape[:2]

kecil = cv.resize(
    gray,
    (int(lebar * 0.5), int(tinggi * 0.5))
)

besar = cv.resize(
    kecil,
    (lebar, tinggi)
)

rata_akhir = np.mean(besar)

selisih = rata_akhir - rata_awal

print("Ukuran awal       :", lebar, "x", tinggi)
print("Ukuran setelah kecil :", kecil.shape[1], "x", kecil.shape[0])
print("Ukuran setelah besar :", besar.shape[1], "x", besar.shape[0])
print("Rata-rata intensitas awal  :", rata_awal)
print("Rata-rata intensitas akhir :", rata_akhir)
print("Perubahan rata-rata        :", selisih)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(gray, cmap='gray')
plt.title("Citra Awal")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(besar, cmap='gray')
plt.title("Setelah Resize 50% lalu Kembali")
plt.axis("off")

plt.show()
```

```python
img1 = cv.imread('/content/drive/MyDrive/PCVK/Gambar/KTP.jpg')
img2 = cv.imread('/content/drive/MyDrive/PCVK/Gambar/foto_aktivitas.jpeg')

nama = "NAMA ANDA"

def label_adaptif(img, nama):

    hasil = img.copy()

    tinggi, lebar = hasil.shape[:2]

    tinggi_bar = int(tinggi * 0.12)
    tebal = max(1, int(min(tinggi, lebar) * 0.006))
    ukuran_font = min(tinggi, lebar) * 0.035

    posisi_x = int(lebar * 0.03)
    posisi_y = int(tinggi - tinggi_bar * 0.30)

    cv.rectangle(
        hasil,
        (0, tinggi - tinggi_bar),
        (lebar, tinggi),
        (0, 0, 0),
        -1
    )

    cv.putText(
        hasil,
        nama,
        (posisi_x, posisi_y),
        cv.FONT_HERSHEY_SIMPLEX,
        ukuran_font,
        (255, 255, 255),
        tebal,
        cv.LINE_AA
    )

    return hasil

hasil1 = label_adaptif(img1, nama)
hasil2 = label_adaptif(img2, nama)

hasil1_rgb = cv.cvtColor(hasil1, cv.COLOR_BGR2RGB)
hasil2_rgb = cv.cvtColor(hasil2, cv.COLOR_BGR2RGB)

plt.figure(figsize=(14, 7))

plt.subplot(1, 2, 1)
plt.imshow(hasil1_rgb)
plt.title(
    "Citra 1: " +
    str(img1.shape[1]) +
    " x " +
    str(img1.shape[0])
)
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(hasil2_rgb)
plt.title(
    "Citra 2: " +
    str(img2.shape[1]) +
    " x " +
    str(img2.shape[0])
)
plt.axis("off")

plt.show()

print("STUDI KASUS 1")
print("Citra 1 :", img1.shape[1], "x", img1.shape[0])
print("Citra 2 :", img2.shape[1], "x", img2.shape[0])
print("Tinggi bar citra 1 :", int(img1.shape[0] * 0.12))
print("Tinggi bar citra 2 :", int(img2.shape[0] * 0.12))
```

```python
def thumbnail_persegi(img, nama_toko):

    tinggi, lebar = img.shape[:2]

    ukuran = max(tinggi, lebar)

    kanvas = np.zeros(
        (ukuran, ukuran, 3),
        dtype=np.uint8
    )

    kanvas[:] = (240, 240, 240)

    if lebar > tinggi:
        tinggi_baru = int(tinggi * ukuran / lebar)
        lebar_baru = ukuran
    else:
        lebar_baru = int(lebar * ukuran / tinggi)
        tinggi_baru = ukuran

    hasil_resize = cv.resize(
        img,
        (lebar_baru, tinggi_baru)
    )

    posisi_x = int((ukuran - lebar_baru) / 2)
    posisi_y = int((ukuran - tinggi_baru) / 2)

    kanvas[
        posisi_y:posisi_y + tinggi_baru,
        posisi_x:posisi_x + lebar_baru
    ] = hasil_resize

    ukuran_logo = ukuran * 0.035
    tebal_logo = max(
        1,
        int(ukuran * 0.004)
    )

    posisi_logo_x = int(ukuran * 0.05)
    posisi_logo_y = int(ukuran * 0.95)

    cv.putText(
        kanvas,
        nama_toko,
        (posisi_logo_x, posisi_logo_y),
        cv.FONT_HERSHEY_SIMPLEX,
        ukuran_logo,
        (0, 0, 0),
        tebal_logo,
        cv.LINE_AA
    )

    return kanvas


nama_toko = "TOKO ANDA"

thumbnail1 = thumbnail_persegi(
    img1,
    nama_toko
)

thumbnail2 = thumbnail_persegi(
    img2,
    nama_toko
)

thumbnail1_rgb = cv.cvtColor(
    thumbnail1,
    cv.COLOR_BGR2RGB
)

thumbnail2_rgb = cv.cvtColor(
    thumbnail2,
    cv.COLOR_BGR2RGB
)

plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(thumbnail1_rgb)
plt.title("Thumbnail Persegi Citra 1")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(thumbnail2_rgb)
plt.title("Thumbnail Persegi Citra 2")
plt.axis("off")

plt.show()

print("\nSTUDI KASUS 2")
print("Ukuran thumbnail 1 :", thumbnail1.shape[1], "x", thumbnail1.shape[0])
print("Ukuran thumbnail 2 :", thumbnail2.shape[1], "x", thumbnail2.shape[0])
```
