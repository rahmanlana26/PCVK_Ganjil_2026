```python
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
import matplotlib.pyplot as plt
import numpy as np
from google.colab import drive
drive.mount('/content/drive')

img = cv.imread('/content/drive/MyDrive/PCVK/Gambar/KTP.jpg')
cv2_imshow(img)
print("Resolusi gambar:", img.shape[1], "x", img.shape[0])
```

```python
img2 = cv.cvtColor(img, cv.COLOR_BGR2RGB)

plt.figure(figsize=(10, 6))
plt.imshow(img2)
plt.show()
```

```python
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

plt.figure(figsize=(10, 6))
plt.imshow(gray, cmap='gray')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.imshow(gray, cmap='magma')
plt.show()
```

```python
img_resize = cv.resize(img2, (347, 229))

plt.figure(figsize=(10, 6))
plt.imshow(img_resize)
plt.show()
```

```python
img_flip = cv.flip(img2, 0)

plt.figure(figsize=(10, 6))
plt.imshow(img_flip)
plt.show()
```

```python
canvas = np.zeros((600, 800, 3), dtype=np.uint8)

h, w = img_flip.shape[:2]

x = (800 - w) // 2
y = (600 - h) // 2

if x >= 0 and y >= 0:
    canvas[y:y+h, x:x+w] = img_flip
else:
    resized = cv.resize(img_flip, (min(w, 800), min(h, 600)))
    rh, rw = resized.shape[:2]
    x = (800 - rw) // 2
    y = (600 - rh) // 2
    canvas[y:y+rh, x:x+rw] = resized

plt.figure(figsize=(10, 7))
plt.imshow(canvas)
plt.show()
```

```python
black_img = np.zeros((530, 530, 3), dtype=np.int16)

# Persegi panjang
cv.rectangle(
    black_img,
    (290, 0),
    (520, 130),
    (0, 0, 255),
    10
)

plt.figure(figsize=(6, 6))
plt.imshow(black_img)
plt.show()

```

```python
cv.rectangle(
    black_img,
    (250, 250),
    (350, 350),
    (255, 0, 0),
    15
)

plt.figure(figsize=(6, 6))
plt.imshow(black_img)
plt.show()
```

```python
cv.circle(
    black_img,
    (150, 150),
    60,
    (0, 255, 0),
    8
)

plt.figure(figsize=(6, 6))
plt.imshow(black_img)
plt.show()
```

```python
cv.circle(
    black_img,
    (400, 400),
    50,
    (0, 255, 0),
    -1
)

cv.line(
    black_img,
    (0, 0),
    (530, 530),
    (255, 255, 255),
    5
)

plt.figure(figsize=(6, 6))
plt.imshow(black_img)
plt.show()
```

```python
cv.putText(
    black_img,
    'PCVK is fun!',
    (10, 500),
    cv.FONT_HERSHEY_SIMPLEX,
    2,
    (255, 255, 0),
    3
)

plt.figure(figsize=(7, 7))
plt.imshow(black_img)
plt.show()
```

```python
black_img2 = np.zeros((530, 530, 3), dtype=np.int32)
vertices = np.array([
    [150, 350],
    [250, 250],
    [450, 350],
    [250, 450]
], dtype=np.int32)

pts = vertices.reshape((-1, 1, 2))
cv.polylines(
    black_img2,
    [pts],
    True,
    (0, 255, 0),
    5
)

plt.figure(figsize=(7, 7))
plt.imshow(black_img2)
plt.show()
```

```python
int16_img = np.zeros((530, 530, 3), dtype=np.int16)
int32_img = np.zeros((530, 530, 3), dtype=np.int32)

print("INT16")
print("dtype  :", int16_img.dtype)
print("min    :", int16_img.min())
print("max    :", int16_img.max())
print("nbytes :", int16_img.nbytes)

print("\nINT32")
print("dtype  :", int32_img.dtype)
print("min    :", int32_img.min())
print("max    :", int32_img.max())
print("nbytes :", int32_img.nbytes)

print("\nPerbandingan memori:")
print("INT32 / INT16 =", int32_img.nbytes / int16_img.nbytes)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(int16_img)
plt.title("INT16")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(int32_img)
plt.title("INT32")
plt.axis("off")

plt.show()
```
